```python
#NumPy Matrix Program

#step 1: Import NumPy
import numpy as np
```

```python
#step 2: Create Matrices
A=np.array([[1, 2, 3],
           [4, 5, 6]])
B=np.array([[7, 8, 9],
            [1, 2, 3]])
C = np.array([[1, 2],
              [3, 4],
              [5, 6]])
D = np.array([[1, 2],
              [3, 4]])
```

```python
#Step 3: Display Matrices
print("----Display Matrices----")
print("Matrix A:\n",A)
print("Matrix B:\n",B)
print("Matrix C:\n",C)
print("Matrix D:\n",D)
```

```python
#Step 4: Matrix Properties
print("----Matrix Properties----\n")
print("Shape: ",A.shape)
print("Dimension: ",A.ndim)
print("Size: ",A.size)
```

```python
#Step 5: Access Elements
print("----Access Elements----\n")
print("A[0,1]:", A[0,1])
print("A[1,2]:", A[1,2])
```

```python
#Step 6: Basic Operations
print("A+B:\n",A+B)
print("A-B:\n",A-B)
print("A*2:\n",A*2)
```

```python
#Step 7: Matrix Multiplication
print("----Matrix Multiplication----")
print("C@A:\n",C @ A)
```

```python
#Step 8: Transpose
print("----Transpose of A----")
print(A.T)
```

```python
# Step 9: Special Matrices
print("----Special Matrices----")
print("Zero Matrix:\n",np.zeros((2,3)))
print("One Matrix:\n",np.ones((2,3)))
print("Identity Matrix:\n",np.eye(4))
```
